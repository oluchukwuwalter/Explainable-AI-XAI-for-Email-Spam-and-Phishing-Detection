```python
# ==========================================================
# CELL 13: HANDLE MISSING VALUES
# ==========================================================

# ----------------------------------------------------------
# FILL MISSING SUBJECTS
# ----------------------------------------------------------
# Some emails have no subject.
# Instead of deleting them, we replace with:
# "No Subject"

clean_ceas_dataset["subject"] = (
    clean_ceas_dataset["subject"]
    .fillna("No Subject")
)


# ----------------------------------------------------------
# FILL MISSING RECEIVERS
# ----------------------------------------------------------
# Replace missing receiver with "Unknown"

clean_ceas_dataset["receiver"] = (
    clean_ceas_dataset["receiver"]
    .fillna("Unknown")
)


# ----------------------------------------------------------
# FILL MISSING URLS
# ----------------------------------------------------------
# Replace missing URLs with empty text

clean_ceas_dataset["urls"] = (
    clean_ceas_dataset["urls"]
    .fillna("")
)


# ----------------------------------------------------------
# VERIFY RESULTS
# ----------------------------------------------------------

print("Missing values handled successfully")


print("\nRemaining Missing Values:")
print(clean_ceas_dataset.isnull().sum())
```

```python
# ==========================================================
# CELL 14: CREATE FINAL EMAIL TEXT COLUMN
# ==========================================================

# ----------------------------------------------------------
# COMBINE SUBJECT + BODY
# ----------------------------------------------------------
# We combine both columns into one single email text

clean_ceas_dataset["email_text"] = (
    
    # Convert subject to text
    clean_ceas_dataset["subject"].astype(str)
    
    # Add space between subject and body
    + " "
    
    # Convert body to text
    + clean_ceas_dataset["body"].astype(str)
)


# ----------------------------------------------------------
# DISPLAY SUCCESS MESSAGE
# ----------------------------------------------------------

print("Final email_text column created successfully")


# ----------------------------------------------------------
# PREVIEW RESULT
# ----------------------------------------------------------

print("\nPreview of Combined Email Text:")

print(
    clean_ceas_dataset[
        ["subject", "body", "email_text"]
    ].head(3)
)
```

```python
# ==========================================================
# CELL 15: CLEAN EMAIL TEXT
# ==========================================================

# ----------------------------------------------------------
# CREATE TEXT CLEANING FUNCTION
# ----------------------------------------------------------
# A function is reusable code.
# This function cleans one email at a time.

def clean_email_text(text):
    
    # Convert to lowercase
    # Example:
    # "HELLO" → "hello"
    text = text.lower()
    
    
    # ------------------------------------------------------
    # REMOVE URLs
    # ------------------------------------------------------
    # Removes:
    # http://
    # https://
    # www.website.com
    text = re.sub(r'http\S+|www\S+', '', text)
    
    
    # ------------------------------------------------------
    # REMOVE EMAIL ADDRESSES
    # ------------------------------------------------------
    text = re.sub(r'\S+@\S+', '', text)
    
    
    # ------------------------------------------------------
    # REMOVE PUNCTUATION
    # ------------------------------------------------------
    # Example:
    # hello!!! → hello
    text = text.translate(
        str.maketrans('', '', string.punctuation)
    )
    
    
    # ------------------------------------------------------
    # REMOVE NUMBERS
    # ------------------------------------------------------
    # Example:
    # win 5000 dollars → win dollars
    text = re.sub(r'\d+', '', text)
    
    
    # ------------------------------------------------------
    # REMOVE EXTRA SPACES
    # ------------------------------------------------------
    text = text.strip()
    
    
    # Return cleaned text
    return text


# ----------------------------------------------------------
# APPLY CLEANING TO ENTIRE DATASET
# ----------------------------------------------------------
# Apply cleaning function to email_text column

clean_ceas_dataset["cleaned_email_text"] = (
    clean_ceas_dataset["email_text"]
    .apply(clean_email_text)
)


# ----------------------------------------------------------
# DISPLAY SUCCESS MESSAGE
# ----------------------------------------------------------

print("Email text cleaned successfully")


# ----------------------------------------------------------
# SHOW BEFORE VS AFTER
# ----------------------------------------------------------

print("\nBEFORE CLEANING:")
print(clean_ceas_dataset["email_text"].iloc[0])

print("\nAFTER CLEANING:")
print(clean_ceas_dataset["cleaned_email_text"].iloc[0])
```

```python
# ==========================================================
# CELL 17: NLP PREPROCESSING
# TOKENISATION + STOPWORD REMOVAL + LEMMATIZATION
# ==========================================================

# ----------------------------------------------------------
# CREATE STOPWORD LIST
# ----------------------------------------------------------

# Load English stopwords
stop_words = set(stopwords.words("english"))

# Create lemmatizer
lemmatizer = WordNetLemmatizer()


# ----------------------------------------------------------
# CREATE NLP PREPROCESSING FUNCTION
# ----------------------------------------------------------

def preprocess_text(text):
    
    # --------------------------------------
    # TOKENIZE TEXT
    # --------------------------------------
    # Split sentence into words
    tokens = word_tokenize(text)
    
    
    # --------------------------------------
    # REMOVE STOPWORDS
    # --------------------------------------
    # Remove common words like:
    # the, is, at, in
    tokens = [
        word for word in tokens
        if word not in stop_words
    ]
    
    
    # --------------------------------------
    # LEMMATIZATION
    # --------------------------------------
    # Convert words to root form
    tokens = [
        lemmatizer.lemmatize(word)
        for word in tokens
    ]
    
    
    # --------------------------------------
    # JOIN WORDS BACK TO SENTENCE
    # --------------------------------------
    processed_text = " ".join(tokens)
    
    return processed_text


# ----------------------------------------------------------
# APPLY NLP PREPROCESSING
# ----------------------------------------------------------

clean_ceas_dataset["processed_email_text"] = (
    clean_ceas_dataset["cleaned_email_text"]
    .apply(preprocess_text)
)


# ----------------------------------------------------------
# DISPLAY SUCCESS MESSAGE
# ----------------------------------------------------------

print("NLP preprocessing completed successfully")


# ----------------------------------------------------------
# SHOW BEFORE VS AFTER
# ----------------------------------------------------------

sample_index = 0

print("\nCLEANED TEXT:")
print(
    clean_ceas_dataset["cleaned_email_text"]
    .iloc[sample_index][:500]
)

print("\nPROCESSED TEXT:")
print(
    clean_ceas_dataset["processed_email_text"]
    .iloc[sample_index][:500]
)
```

## Step 3.3: Develop a Reusable Email Parsing Function

### Objective

The objective of this step is to develop a reusable email parsing function capable of extracting structured information from raw email messages. The parser will isolate relevant email components, including the subject, sender, recipient, date, and message body, while handling multipart emails, encoded content, and missing fields. This reusable function will subsequently be applied to both the Enron and SpamAssassin benchmark datasets, ensuring consistency throughout the data engineering pipeline.

```python
# ==========================================================
# STEP 3.3 : IMPORT EMAIL PARSING LIBRARIES
# ==========================================================

import email
from email import policy
from email.parser import Parser

import pandas as pd
```

```python
# ==========================================================
# STEP 3.3 : EMAIL PARSER
# ==========================================================

def parse_email(raw_message):
    """
    Extract useful information from a raw email message.
    """

    try:

        msg = Parser(policy=policy.default).parsestr(raw_message)

        subject = msg.get("Subject", "")
        sender = msg.get("From", "")
        recipient = msg.get("To", "")
        date = msg.get("Date", "")

        body = ""

        if msg.is_multipart():

            for part in msg.walk():

                if part.get_content_type() == "text/plain":

                    try:
                        body += part.get_content()

                    except Exception:
                        pass

        else:

            try:
                body = msg.get_content()

            except Exception:
                body = str(msg.get_payload())

        return pd.Series({

            "subject": subject,

            "from": sender,

            "to": recipient,

            "date": date,

            "body": body

        })

    except Exception:

        return pd.Series({

            "subject": "",

            "from": "",

            "to": "",

            "date": "",

            "body": ""

        })
```

## Step 3.5: Develop a Chunk-Based Processing Pipeline

### Objective

The objective of this step is to develop a scalable preprocessing pipeline capable of processing the complete Enron Email Dataset in manageable chunks. Since the dataset is approximately 1.32 GB, loading it entirely into memory is inefficient and may exceed available system resources. Therefore, chunk-based processing will be implemented to improve memory efficiency while ensuring that every email undergoes the same parsing, standardization, and validation procedures before being merged into a single processed dataset.

```python
# ==========================================================
# STANDARD DATASET SCHEMA
# ==========================================================

STANDARD_COLUMNS = [
    "dataset",
    "subject",
    "body",
    "email_text",
    "label"
]

print(STANDARD_COLUMNS)
```

```python
# ==========================================================
# STANDARDIZE ENRON DATASET
# ==========================================================

def standardize_enron(parsed_df):
    """
    Convert parsed Enron emails into the common dataset schema.
    """

    df = parsed_df.copy()

    df["dataset"] = "Enron"

    # Enron contains legitimate corporate emails
    df["label"] = 0

    df["subject"] = df["subject"].fillna("")
    df["body"] = df["body"].fillna("")

    df["email_text"] = (
        df["subject"] + " " + df["body"]
    )

    return df[STANDARD_COLUMNS]
```

## Step 3.9: Develop the Enron Processing Pipeline

### Objective

The objective of this step is to integrate all preprocessing components into a single automated pipeline capable of processing the complete Enron Email Dataset. The pipeline sequentially reads the dataset in manageable chunks, parses raw email messages, standardizes the extracted information, validates data quality, records processing statistics, and saves the processed output to disk. This modular design improves scalability, reproducibility, and fault tolerance while supporting the downstream stages of the hybrid phishing detection framework.

```python
# ==========================================================
# COMPLETE ENRON PROCESSING PIPELINE
# ==========================================================

import time

def process_enron_dataset(
    input_file,
    output_file,
    chunk_size=25000
):
    """
    Process the complete Enron dataset using chunk-based processing.
    """

    # Reset master statistics
    master_stats = {
        "chunks": 0,
        "rows": 0,
        "duplicates": 0,
        "missing_subject": 0,
        "missing_body": 0,
        "missing_email_text": 0
    }

    # Remove existing output file
    if output_file.exists():
        output_file.unlink()

    start_time = time.time()

    for chunk_number, chunk in enumerate(
        pd.read_csv(
            input_file,
            chunksize=chunk_size,
            low_memory=False
        ),
        start=1
    ):

        chunk_start = time.time()

        # Process the chunk
        processed_chunk = process_chunk(chunk)

        # Validate
        validation = validate_chunk(processed_chunk)

        # Save
        processed_chunk.to_csv(
            output_file,
            mode="a",
            header=not output_file.exists(),
            index=False
        )

        # Update statistics
        update_master_stats(master_stats, validation)

        # Log progress
        elapsed = time.time() - chunk_start

        log_chunk(
            chunk_number,
            validation,
            elapsed
        )

    total_time = time.time() - start_time

    return master_stats, total_time
```

# Phase 0.3: Utility Functions

## Email Parsing Function

The following utility function extracts the subject and body from raw email messages. The function is reusable across the Enron and SpamAssassin benchmark datasets, ensuring consistent preprocessing throughout the data engineering pipeline.

```python
# ==========================================================
# UTILITY FUNCTION : PARSE EMAIL
# ==========================================================

from email import policy
from email.parser import Parser

def parse_email(raw_email):
    """
    Extract the subject and body from a raw email message.
    Returns:
        subject (str)
        body (str)
    """

    try:
        message = Parser(policy=policy.default).parsestr(raw_email)

        subject = message.get("Subject", "")

        if message.is_multipart():

            body_parts = []

            for part in message.walk():

                if part.get_content_type() == "text/plain":

                    try:
                        body_parts.append(part.get_content())

                    except Exception:
                        pass

            body = "\n".join(body_parts)

        else:

            body = message.get_content()

        return subject, body

    except Exception:

        return "", raw_email
```

## Step 8.3: Remove HTML Tags

### Objective

The objective of this step is to remove HTML markup while preserving the visible textual content of each email. This reduces formatting noise without affecting the semantic information required for Transformer models.

```python
from bs4 import BeautifulSoup
import pandas as pd
import re

def remove_html(text):

    if pd.isna(text):
        return ""

    text = str(text)

    # Only clean if genuine HTML tags are present
    if re.search(
        r"</?(html|body|div|span|p|br|table|tr|td|a|img|head|style|script)\b",
        text,
        flags=re.IGNORECASE
    ):
        return BeautifulSoup(
            text,
            "html.parser"
        ).get_text(separator=" ")

    return text
```

```python
master_df["cleaned_email_text"] = (
    master_df["cleaned_email_text"]
    .apply(remove_html)
)
```

```python
# ==========================================================
# STEP 8.4 : HTML ENTITY DECODING
# ==========================================================

import html

master_df["cleaned_email_text"] = (

    master_df["cleaned_email_text"]

    .apply(html.unescape)

)
```

```python
# ==========================================================
# STEP 8.5 : REMOVE NON-PRINTABLE CHARACTERS
# ==========================================================

import re

def remove_nonprintable(text):

    return re.sub(

        r"[\x00-\x1F\x7F]",

        " ",

        str(text)

    )

master_df["cleaned_email_text"] = (

    master_df["cleaned_email_text"]

    .apply(remove_nonprintable)

)
```

```python
# ==========================================================
# STEP 8.6 : NORMALIZE WHITESPACE
# ==========================================================

master_df["cleaned_email_text"] = (

    master_df["cleaned_email_text"]

    .str.replace(

        r"\s+",

        " ",

        regex=True

    )

    .str.strip()

)
```

## Step 8.7: Unicode Normalization

### Objective

The objective of this step is to normalize Unicode characters into a consistent representation while preserving the semantic meaning of the email text. This improves compatibility with downstream Transformer tokenization.

```python
# ==========================================================
# STEP 8.7 : UNICODE NORMALIZATION
# ==========================================================

import unicodedata

def normalize_unicode(text):

    return unicodedata.normalize(
        "NFKC",
        str(text)
    )

master_df["cleaned_email_text"] = (
    master_df["cleaned_email_text"]
    .apply(normalize_unicode)
)
```

## Step 9.6: Batch Tokenization and Incremental Export

### Objective

The objective of this step is to tokenize the cleaned email corpus in manageable batches and save each batch immediately to disk. This approach minimizes memory consumption, improves scalability, and creates reusable Transformer-ready inputs for downstream contextual embedding generation.

```python
# ==========================================================
# STEP 9.6A : CREATE OUTPUT FOLDER
# ==========================================================

from pathlib import Path
import torch

tokenized_dir = PROCESSED_DATA / "tokenized_batches"

tokenized_dir.mkdir(
    parents=True,
    exist_ok=True
)

print("=" * 70)
print("TOKENIZED OUTPUT DIRECTORY")
print("=" * 70)
print(tokenized_dir)
```

```python
# ==========================================================
# STEP 9.6B : TOKENIZE AND SAVE BATCHES
# ==========================================================

BATCH_SIZE = 1000

total_rows = len(token_df)

batch_number = 1

print("=" * 70)
print("BATCH TOKENIZATION STARTED")
print("=" * 70)

for start in range(0, total_rows, BATCH_SIZE):

    end = min(start + BATCH_SIZE, total_rows)

    batch = token_df.iloc[start:end]

    encoded = tokenizer(

        batch["cleaned_email_text"].tolist(),

        padding="max_length",

        truncation=True,

        max_length=512,

        return_attention_mask=True,

        return_tensors="pt"

    )

    batch_data = {

        "input_ids": encoded["input_ids"],

        "attention_masks": encoded["attention_mask"],

        "labels": torch.tensor(
            batch["label"].values,
            dtype=torch.long
        )

    }

    batch_file = tokenized_dir / f"batch_{batch_number:03d}.pt"

    torch.save(
        batch_data,
        batch_file
    )

    print(
        f"Batch {batch_number:03d} saved "
        f"({start:,} - {end-1:,})"
    )

    batch_number += 1

print("=" * 70)
print("TOKENIZATION COMPLETED")
print("=" * 70)
```

## Experiment 10.6: Transformer Tokenization Evaluation

### Objective

The objective of this experiment is to evaluate the Transformer compatibility of the cleaned email corpus by analysing token sequence lengths generated by the BERT tokenizer. The evaluation measures token length statistics, truncation requirements, and sequence distributions to determine the suitability of the dataset for contextual embedding generation.

```python
# ==========================================================
# EXPERIMENT 10.6
# TOKEN LENGTH ANALYSIS
# ==========================================================

from tqdm import tqdm

print("=" * 70)
print("CALCULATING TOKEN LENGTHS")
print("=" * 70)

token_lengths = []

for text in tqdm(master_df["cleaned_email_text"]):

    tokens = tokenizer.tokenize(str(text))

    token_lengths.append(len(tokens))

master_df["token_length"] = token_lengths

print("Completed.")
```
