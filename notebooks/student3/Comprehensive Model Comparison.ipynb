```python
NOTEBOOK_DIR = Path.cwd()

PROJECT_ROOT = NOTEBOOK_DIR.parent.parent

REPORTS = PROJECT_ROOT / "reports"

FIGURES = PROJECT_ROOT / "figures"

print("="*70)
print("MODEL COMPARISON")
print("="*70)
```

```python
bert_summary = pd.read_csv(

    REPORTS /

    "bert_performance_summary.csv"

)

roberta_summary = pd.read_csv(

    REPORTS /

    "roberta_performance_summary.csv"

)

bert_xgb_summary = pd.read_csv(

    REPORTS /

    "bert_xgboost_performance_summary.csv"

)

roberta_xgb_summary = pd.read_csv(

    REPORTS /

    "roberta_xgboost_performance_summary.csv"

)
```

```python
bert_report = pd.read_csv(

    REPORTS /

    "bert_classification_report.csv",

    index_col=0

)

roberta_report = pd.read_csv(

    REPORTS /

    "roberta_classification_report.csv",

    index_col=0

)

bert_xgb_report = pd.read_csv(

    REPORTS /

    "bert_xgboost_classification_report.csv",

    index_col=0

)

roberta_xgb_report = pd.read_csv(

    REPORTS /

    "roberta_xgboost_classification_report.csv",

    index_col=0

)
```

```python
comparison = pd.DataFrame({

    "Model":[

        "BERT",

        "RoBERTa",

        "BERT + XGBoost",

        "RoBERTa + XGBoost"

    ],

    "Accuracy":[

        0.9940,

        0.9930,

        0.9920,

        0.9920

    ],

    "Spam Precision":[

        0.9773,

        0.9663,

        0.9659,

        0.9659

    ],

    "Spam Recall":[

        0.9556,

        0.9556,

        0.9444,

        0.9444

    ],

    "Spam F1":[

        0.9663,

        0.9609,

        0.9551,

        0.9551

    ],

    "Balanced Accuracy":[

        0.9767,

        0.9761,

        0.9706,

        0.9706

    ],

    "ROC-AUC":[

        None,

        None,

        0.9917,

        0.9976

    ],

    "PR-AUC":[

        None,

        None,

        0.9811,

        0.9834

    ]

})

comparison
```

```python
comparison.to_csv(

    REPORTS /

    "overall_model_comparison.csv",

    index=False

)

print("="*70)
print("MASTER COMPARISON SAVED")
print("="*70)
```

```python
plt.figure(figsize=(8,6))

plt.bar(

    comparison["Model"],

    comparison["Accuracy"]

)

plt.ylabel("Accuracy")

plt.title("Accuracy Comparison")

plt.xticks(rotation=20)

plt.tight_layout()

plt.savefig(

    FIGURES /

    "accuracy_comparison.png",

    dpi=300

)

plt.show()
```

```python
plt.figure(figsize=(8,6))

plt.bar(

    comparison["Model"],

    comparison["Spam F1"]

)

plt.ylabel("Spam F1")

plt.title("Spam F1 Comparison")

plt.xticks(rotation=20)

plt.tight_layout()

plt.savefig(

    FIGURES /

    "spam_f1_comparison.png",

    dpi=300

)

plt.show()
```

```python
ranking = comparison.sort_values(

    by="Accuracy",

    ascending=False

)

display(ranking)

ranking.to_csv(

    REPORTS /

    "model_ranking.csv",

    index=False

)
```
