```python
# ==========================================================
# STUDENT 3
# NOTEBOOK 07
# AUTOMATED MODEL COMPARISON
# ==========================================================

from pathlib import Path

import pandas as pd

import matplotlib.pyplot as plt
```

```python
NOTEBOOK_DIR = Path.cwd()

PROJECT_ROOT = NOTEBOOK_DIR.parent.parent

REPORTS = PROJECT_ROOT / "reports"

FIGURES = PROJECT_ROOT / "figures"

print("="*70)
print("AUTOMATED MODEL COMPARISON")
print("="*70)
```

```python
bert_summary = pd.read_csv(
    REPORTS / "bert_performance_summary.csv"
)

roberta_summary = pd.read_csv(
    REPORTS / "roberta_performance_summary.csv"
)

bert_xgb_summary = pd.read_csv(
    REPORTS / "bert_xgboost_performance_summary.csv"
)

roberta_xgb_summary = pd.read_csv(
    REPORTS / "roberta_xgboost_performance_summary.csv"
)
```

```python
bert_report = pd.read_csv(
    REPORTS / "bert_classification_report.csv",
    index_col=0
)

roberta_report = pd.read_csv(
    REPORTS / "roberta_classification_report.csv",
    index_col=0
)

bert_xgb_report = pd.read_csv(
    REPORTS / "bert_xgboost_classification_report.csv",
    index_col=0
)

roberta_xgb_report = pd.read_csv(
    REPORTS / "roberta_xgboost_classification_report.csv",
    index_col=0
)
```

```python
def metric(summary, name):

    return summary.loc[
        summary["Metric"] == name,
        "Value"
    ].values[0]
```

```python
comparison = pd.DataFrame({

    "Model":[

        "BERT",

        "RoBERTa",

        "BERT + XGBoost",

        "RoBERTa + XGBoost"

    ],

    "Accuracy":[

        metric(bert_summary,"Accuracy"),

        metric(roberta_summary,"Accuracy"),

        metric(bert_xgb_summary,"Accuracy"),

        metric(roberta_xgb_summary,"Accuracy")

    ],

    "Balanced Accuracy":[

        metric(bert_summary,"Balanced Accuracy"),

        metric(roberta_summary,"Balanced Accuracy"),

        metric(bert_xgb_summary,"Balanced Accuracy"),

        metric(roberta_xgb_summary,"Balanced Accuracy")

    ],

    "ROC-AUC":[

        metric(bert_summary,"ROC-AUC"),

        metric(roberta_summary,"ROC-AUC"),

        metric(bert_xgb_summary,"ROC-AUC"),

        metric(roberta_xgb_summary,"ROC-AUC")

    ],

    "PR-AUC":[

        metric(bert_summary,"PR-AUC"),

        metric(roberta_summary,"PR-AUC"),

        metric(bert_xgb_summary,"PR-AUC"),

        metric(roberta_xgb_summary,"PR-AUC")

    ],

    "MCC":[

        metric(bert_summary,"Matthews Correlation Coefficient"),

        metric(roberta_summary,"Matthews Correlation Coefficient"),

        metric(bert_xgb_summary,"Matthews Correlation Coefficient"),

        metric(roberta_xgb_summary,"Matthews Correlation Coefficient")

    ],

    "Kappa":[

        metric(bert_summary,"Cohen's Kappa"),

        metric(roberta_summary,"Cohen's Kappa"),

        metric(bert_xgb_summary,"Cohen's Kappa"),

        metric(roberta_xgb_summary,"Cohen's Kappa")

    ],

    "Spam Precision":[

        bert_report.loc["1","precision"],

        roberta_report.loc["1","precision"],

        bert_xgb_report.loc["1","precision"],

        roberta_xgb_report.loc["1","precision"]

    ],

    "Spam Recall":[

        bert_report.loc["1","recall"],

        roberta_report.loc["1","recall"],

        bert_xgb_report.loc["1","recall"],

        roberta_xgb_report.loc["1","recall"]

    ],

    "Spam F1":[

        bert_report.loc["1","f1-score"],

        roberta_report.loc["1","f1-score"],

        bert_xgb_report.loc["1","f1-score"],

        roberta_xgb_report.loc["1","f1-score"]

    ]

})

comparison
```
